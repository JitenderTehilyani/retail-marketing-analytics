```python
import pandas as pd

# Load the raw data from our folder
df = pd.read_csv("../data/raw/retail_sales_data.csv")

# Look at the first 5 rows of our data
df.head()
```

```python
# Check for empty/blank spaces in our columns
print("Missing values in each column:")
print(df.isnull().sum())

# Check data types (numbers vs dates vs text)
print("\nData columns and their types:")
print(df.dtypes)
```

```python
# 1. Convert the 'Date' column from text (str) into a real computer date
df['Date'] = pd.to_datetime(df['Date'])

# 2. Safety Check: Make sure there are no weird negative numbers in Sales or Quantity
df = df[(df['Total Amount'] > 0) & (df['Quantity'] > 0)]

# 3. Save this clean data into your processed folder
df.to_csv("../data/processed/cleaned_retail_sales.csv", index=False)

print("Data successfully cleaned and saved for your specific dataset!")
```
